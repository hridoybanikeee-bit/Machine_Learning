```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/kaggle/input/datasets/hridoybanik/mushroom-data-set/mushrooms.csv")
```

```python
df.head(5)
```

```python
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
```

```python
print(categorical_cols)
```

```python
categorical_cols = ['cap-shape', 'cap-surface', 'cap-color', 'bruises', 'odor', 'gill-attachment', 'gill-spacing', 'gill-size', 'gill-color', 'stalk-shape', 'stalk-root', 'stalk-surface-above-ring', 'stalk-surface-below-ring', 'stalk-color-above-ring', 'stalk-color-below-ring', 'veil-type', 'veil-color', 'ring-number', 'ring-type', 'spore-print-color', 'population', 'habitat']
```

```python
X = df.drop('class', axis = 1)
y = df['class']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn import svm
```

```python
p1 = Pipeline(
    steps = [
        ("OHE", OneHotEncoder(drop='first',sparse_output=False))
    ]
)
```

```python
preprocessing = ColumnTransformer(
    transformers = [
        ("preprocess", p1, categorical_cols)
    ]
)
```

```python
model_training = Pipeline(
    steps = [
        ("preprocess", preprocessing),
        ("model", svm.SVC())
    ]
)
```

```python
model_training.fit(X_train, y_train)
```

```python
y_train
```

```python
y_test
```

```python
y_pred = model_training.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, f1_score
print("Testing Accuracy")
# Calculate metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred) 
f1 = f1_score(y_test, y_pred)

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"F1 Score:  {f1:.4f}")
```

```python
print("Training Accuracy")
y_train_val = model_training.predict(X_train)
accuracy = accuracy_score(y_train, y_train_val)
precision = precision_score(y_train, y_train_val) 
f1 = f1_score(y_train, y_train_val)

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"F1 Score:  {f1:.4f}")
```
