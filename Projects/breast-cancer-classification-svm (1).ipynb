```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/kaggle/input/datasets/hridoybanik/breast-can-classification/data.csv")
```

```python
df.head(5)
```

```python
df.drop('id', axis = 1, inplace = True)
```

```python
df['diagnosis'].value_counts()
```

```python
df.drop('Unnamed: 32', axis = 1, inplace = True)
```

```python
X = df.drop("diagnosis", axis = 1)
y = df['diagnosis']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()
```

```python
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn import svm
```

```python
p1 = Pipeline(
    steps = [
        ("scaling", StandardScaler()),
        
    ]
        
    
)
```

```python
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
preprocessing = ColumnTransformer(
    transformers = [
        ("scale", p1, numerical_cols)
    ], remainder = 'passthrough'
)
```

```python
model_training = Pipeline(
    steps = [
        ("preprocessings", preprocessing),
        ("model", svm.SVC(kernel = "linear"))
    ]
)
```

```python
model_training.fit(X_train,y_train)
```

```python
y_pred = model_training.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, f1_score
print("Testing Accuracy")
# Calculate metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred) 
f1 = f1_score(y_test, y_pred)

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"F1 Score:  {f1:.4f}")
```

```python
print("Training Accuracy")
y_train_val = model_training.predict(X_train)
accuracy = accuracy_score(y_train, y_train_val)
precision = precision_score(y_train, y_train_val) 
f1 = f1_score(y_train, y_train_val)

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"F1 Score:  {f1:.4f}")
```
