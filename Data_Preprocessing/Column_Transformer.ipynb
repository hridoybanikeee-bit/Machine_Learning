```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv("/content/covid_toy.csv")
df.head(5)
```

```python
X = df.drop('has_covid', axis = 1)
y = df['has_covid']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.compose import ColumnTransformer
transformer = ColumnTransformer(transformers=[
    ('imput',SimpleImputer(),['fever']),
    ('ordinal',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('onehot',OneHotEncoder(sparse_output =False,drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
X_train_transformed = transformer.fit_transform(X_train)
X_test_transformed = transformer.transform(X_test)
```

```python
# Get column names
columns = transformer.get_feature_names_out()

# Convert to DataFrame
X_train_df = pd.DataFrame(X_train_transformed, columns=columns)
X_test_df = pd.DataFrame(X_test_transformed, columns=columns)
```
