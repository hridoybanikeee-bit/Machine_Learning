```python
X = df.drop('target',axis = 1)
y = df['target']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.preprocessing import OneHotEncoder
cp_ohe = OneHotEncoder(sparse_output=False, drop='first').set_output(transform='pandas')
cp_ohe.fit(X_train[['cp']])
cp_df = cp_ohe.transform(X_train[['cp']])
X_train = pd.concat([X_train,cp_df],axis =1)
cp_df = cp_ohe.transform(X_test[['cp']])
X_test = pd.concat([X_test,cp_df],axis =1)
```

```python
thal_ohe = OneHotEncoder(sparse_output=False, drop='first').set_output(transform='pandas')
thal_ohe.fit(X_train[['thal']])
thal_df = thal_ohe.transform(X_train[['thal']])
X_train = pd.concat([X_train,thal_df],axis =1)
thal_df = thal_ohe.transform(X_test[['thal']])
X_test = pd.concat([X_test,thal_df],axis =1)
```
