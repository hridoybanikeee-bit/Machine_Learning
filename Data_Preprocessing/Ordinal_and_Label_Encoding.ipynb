```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv("/content/customer.csv")
```

```python
df.head(5)
```

```python
df = df.iloc[:,2:]
```

```python
df.head(5)
```

```python
X= df.drop('purchased', axis =1)
y = df['purchased']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)
```

```python
X_train['education'].value_counts()
```

```python
X_train['review'].value_counts()
```

```python

from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories = [["Poor","Average","Good"], ['School', 'UG', 'PG']] )
oe.fit(X_train)
X_train_scaled = oe.transform(X_train)
X_test_scaled = oe.transform(X_test)
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled, columns = X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns = X_test.columns)
```

```python
X_train_scaled.head(5)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
le.fit(y_train)
y_train_scaled = le.transform(y_train)
y_test_scaled = le.transform(y_test)
```

```python
y_train_scaled = pd.DataFrame(y_train_scaled, columns = ['purchased'])
y_test_scaled = pd.DataFrame(y_test_scaled, columns = ['purchased'])
```
